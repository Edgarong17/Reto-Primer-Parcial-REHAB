# Selecciona una de las actividades y toma una de las repeticiones. Selecciona uno de los canales del sensor y grafica sus valores en el tiempo.

```python
import matplotlib.pyplot as plt

sensor2 = np.load('Rehab_exercise/d02_processed_data/000_2.npy')

# Frecuencia de muestreo
fs = 50

n_muestras = sensor2.shape[1]
tiempo = np.arange(n_muestras) / fs

una_repeticion = sensor2[0, :, 0]

plt.plot(
    tiempo,
    una_repeticion,
)

plt.xlabel("Tiempo (s)")
plt.ylabel("Thumb flexion angle. (°)")
plt.title("Thumb flexion angle.")
plt.grid(True)

plt.tight_layout()
plt.show()
```

# ¿Se ven balanceados los conteos?
El conteo esta mas o menos balanceado teniendo una media de 288.5 y una desviacion estadar de 44

```python
import numpy as np
import matplotlib.pyplot as plt

nombres_movimientos = [
    "Bobath handshake",
    "Bobath flexion/extension",
    "Bobath forward flexion/extension",
    "Bobath anterior/posterior rotation",
    "Elbow flexion and wrist compression",
    "Wrist flexion and extension",
    "Finger-to-finger training",
    "Ball gripping",
    "Shoulder joint internal and external",
    "Breast expansion",
    "Flexion-pressure rotation forward and backward",
    "Elbow joint flexion and touch",
    "Shoulder touch training",
    "Ankle extension & Knee internal and external rotation",
    "Knee flexion and extension",
    "Hip flexion and extension"
]

nombres_archivos = []
repeticiones = []

for archivo in carpeta.iterdir():

    if archivo.is_file() and archivo.name.endswith("_2.npy"):

        data = np.load(archivo)

        num_repeticiones = data.shape[0]

        nombres_archivos.append(archivo.name)
        repeticiones.append(num_repeticiones)

        print(
            f"Archivo: {archivo.name}, "
            f"Número de repeticiones: {num_repeticiones}"
        )

#Media y desviación estándar del número de repeticiones
media_repeticiones = np.mean(repeticiones)
desviacion_estandar_repeticiones = np.std(repeticiones)

print(f"Media de repeticiones: {media_repeticiones}")
print(f"Desviación estándar de repeticiones: {desviacion_estandar_repeticiones}")

# Gráfica
plt.figure(figsize=(10, 6))

plt.bar(nombres_movimientos, repeticiones)

plt.xlabel("Archivos")
plt.ylabel("Número de repeticiones")
plt.title("Número de repeticiones por archivo")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

# Elige un canal del sensor y calcula la media y mediana de todos los valores de una actividad en específico. Haz esto para cada actividad y genera una tabla con 16 filas y dos columnas. Las filas son la actividad y las columnas sus valores de meda y mediana. Puedes concatenar todas las repeticiones de una actividad y calcular los estadísticos sobre los datos concatenados.

```python
import pandas as pd

datos_sensor2_canal1 = []

for archivo in carpeta.iterdir():

    if archivo.is_file() and archivo.name.endswith("_2.npy"):

        data = np.load(archivo)

        media = np.mean(data[:, :, 0])  # Media del número de repeticiones
        mediana = np.median(data[:, :, 0])  # Mediana del primer valor de cada repetición

        datos_sensor2_canal1.append((media, mediana))
        
df = pd.DataFrame(datos_sensor2_canal1, columns=["Media", "Mediana"])

df.index = nombres_movimientos

print(df)
```

# Elige dos actividades diferentes y grafica su histograma (de la concatenación de todas las repeticiones) del canal seleccionado. Incluye los dos histogramas en la misma gráfica para ver que tanto se traslapan.

# Agrega una línea vertical sobre cada histograma que represente la media de cada actividad.

```python
mov1 = np.load('Rehab_exercise/d02_processed_data/000_2.npy')
mov2 = np.load('Rehab_exercise/d02_processed_data/001_2.npy')

print("Forma de mov1:", mov1.shape)
print("Forma de mov2:", mov2.shape)


#Grafica ambos movimientos en un histograma usando el primer canal de cada movimiento (canal 0) y la media de cada repetición
plt.figure(figsize=(10, 6))



plt.hist(mov1[:,:, 0], bins=20, alpha=0.5, label='Movimiento 1', color='blue')
plt.hist(mov2[:,:, 0], bins=20, alpha=0.5, label='Movimiento 2', color='orange')

plt.xlabel("Valor del primer canal")
plt.ylabel("Frecuencia")
plt.title("Histograma de los primeros canales de dos movimientos")
plt.legend()


#Media de cada actividad
media_mov1_chc1 = np.mean(mov1[:,:, 0])
media_mov2_chc1 = np.mean(mov2[:,:, 0])

#Agregar linea vertical en la media de cada actividad
plt.axvline(media_mov1_chc1, color='blue', linestyle='dashed', linewidth=2, label='Media Movimiento 1')
plt.axvline(media_mov2_chc1, color='orange', linestyle='dashed', linewidth=2, label='Media Movimiento 2')

print(f"Media del primer canal del Movimiento 1: {media_mov1_chc1}")
print(f"Media del primer canal del Movimiento 2: {media_mov2_chc1}")
```

```python
mov1 = np.load('Rehab_exercise/d02_processed_data/000_2.npy')
mov2 = np.load('Rehab_exercise/d02_processed_data/001_2.npy')

print("Forma de mov1:", mov1.shape)
print("Forma de mov2:", mov2.shape)

conc_mov1 = np.mean(mov1, axis=0)
conc_mov2 = np.mean(mov2, axis=0)

print("Forma de conc_mov1:", conc_mov1.shape)
print("Forma de conc_mov2:", conc_mov2.shape)

#Grafica ambos movimientos en un histograma usando el primer canal de cada movimiento (canal 0) y la media de cada repetición
plt.figure(figsize=(10, 6))

plt.hist(conc_mov1[:, 1], bins=20, alpha=0.5, label='Movimiento 1', color='blue')
plt.hist(conc_mov2[:, 1], bins=20, alpha=0.5, label='Movimiento 2', color='orange')

plt.xlabel("Valor del primer canal")
plt.ylabel("Frecuencia")
plt.title("Histograma de los primeros canales de dos movimientos")
plt.legend()


#Media de cada actividad
media_mov1_chc2 = np.mean(conc_mov1[:, 1])
media_mov2_chc2 = np.mean(conc_mov2[:, 1])

#Agregar linea vertical en la media de cada actividad
plt.axvline(media_mov1_chc2, color='blue', linestyle='dashed', linewidth=2, label='Media Movimiento 1')
plt.axvline(media_mov2_chc2, color='orange', linestyle='dashed', linewidth=2, label='Media Movimiento 2')

print(f"Media del primer canal del Movimiento 1: {media_mov1_chc2}")
print(f"Media del primer canal del Movimiento 2: {media_mov2_chc2}")
```
